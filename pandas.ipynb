```python
s = pd.Series([10,25,30,45,60,75,90,-5,100])
```

```python
print(s[s>50])
print(s[(s>20)&(s<80)])
print(s[s%2!=0])
print(s[s<0])
# print(s[s].filter(regex='^[1-9][0-9]*$'))
print(s[s!=100])
print(s[(s<20) | (s>80)])
print(s[s%5==0])
```
